# Phase 4 – Exploratory Data Analysis (EDA) :- 

```python
total_students = len(df) 

print("Total Number of Students:", total_students)
```

```python
avg_attendance = df["Attendance"].mean()

avg_cgpa = df["CGPA"].mean()

print("Average Attendance:", round(avg_attendance, 2))
print("Average CGPA:", round(avg_cgpa, 2))
```

```python

highest_marks = df["Total_Marks"].max()
lowest_marks = df["Total_Marks"].min()

print("Highest Total Marks :", highest_marks)
print("Lowest Total Marks :", lowest_marks)
```

```python
dept_cgpa = df.groupby("Department")["CGPA"].mean()

print(dept_cgpa)
```

```python
gender_performance = df.groupby("Gender")["CGPA"].mean()

print(gender_performance)
```

```python
top_stu = df.sort_values(by="CGPA", ascending=False)

print(top_stu.head(10))
```

```python
correlation = df.corr(numeric_only=True)

print(correlation)
```

# Phase 5 – Data Visualization (Matplotlib & Seaborn) :-

```python
dept_cgpa = df.groupby("Department")["CGPA"].mean()

# Create bar chart :- 

plt.figure(figsize=(8,5))
dept_cgpa.plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Average CGPA by Department")
plt.xlabel("Department")
plt.ylabel("Average CGPA")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Attendance"],
         bins=10,
         color="lightgreen",
         edgecolor="black")

plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")

plt.show()
```

```python
gender_count = df["Gender"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Study_Hours"],
    df["CGPA"]
)

plt.title("Study Hours vs CGPA")
plt.xlabel("Study Hours")
plt.ylabel("CGPA")

plt.show()
```

### Each dot represents one student.

- X-axis → Study Hours
- Y-axis → CGPA

```python
correlation = df.corr(numeric_only=True)

plt.figure(figsize=(10,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

## Explanation :-
- corr() calculates the correlation matrix.
- heatmap() displays it as a colored table.
- annot=True shows correlation values inside each cell.
- cmap="coolwarm" colors strong positive and negative correlations differently.
